```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("upi_clean.csv")
df["date_clean"] = pd.to_datetime(df["date_clean"])

analysis = df[(df["is_refund"] == False) & (df["is_outlier"] == False)].copy()
```

```python
analysis["month"] = analysis["date_clean"].dt.to_period("M").astype(str)
monthly = analysis.groupby("month")["amount_clean"].sum()

plt.figure(figsize=(14, 4))
plt.plot(monthly.index, monthly.values, color="#16A34A", marker="o")
plt.xticks(rotation=45)
plt.title("Monthly Spending Trend", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig("chart_trend.png", dpi=150)
```

```python
plt.figure(figsize=(14, 6))
sns.boxplot(data=analysis, x="category_clean", y="amount_clean")
plt.xticks(rotation=45)
plt.title("Amount Distribution by Category")
plt.tight_layout()
plt.savefig("chart_boxplot.png", dpi=150)
```
