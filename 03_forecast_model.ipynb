```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

df = pd.read_csv("upi_clean.csv")
df["date_clean"] = pd.to_datetime(df["date_clean"])

analysis = df[(df["is_refund"] == False) & (df["is_outlier"] == False)].copy()
analysis["month"] = analysis["date_clean"].dt.to_period("M").astype(str)
```

```python
monthly = analysis.groupby("month")["amount_clean"].sum().reset_index()
monthly["month_index"] = range(len(monthly))
```

```python
X = monthly[["month_index"]]
y = monthly["amount_clean"]
model = LinearRegression()
model.fit(X, y)
r2 = model.score(X, y)
print(f"R2 score: {r2:.3f}")
```
